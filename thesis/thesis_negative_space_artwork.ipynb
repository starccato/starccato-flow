# Negative Space Artwork: Galactic Supernova Distribution

This notebook demonstrates the creation of an artistic, ink-drawing style visualization of the galactic supernova distribution. Instead of showing where supernovae ARE, it shows where they are NOT by placing small black dots everywhere except at supernova locations, with feathered edges for an organic appearance.

## 1. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from starccato_flow.supernovae.supernovae import Supernovae
```

## 2. Load Supernova Data

```python
# Initialize the Supernovae class with complex galactic distribution
print("Loading supernova data...")
sn = Supernovae(complex=True, limit=None)  # Load all supernovae
print(f"✓ Loaded {len(sn)} supernovae locations")
```

## 3. Create Negative Space Artwork

```python
# Generate the negative space artwork - CIRCULAR VERSION (recommended for thesis)
print("Creating negative space artwork (circular)...")
print("- Sampling candidate dot positions...")
print("- Filtering out positions near supernovae...")
print("- Applying smooth edge feathering...")

# Create output directory if it doesn't exist
output_dir = Path("/Users/tarineccleston/Desktop/starccato/starccato-flow/outdir")
output_dir.mkdir(parents=True, exist_ok=True)

# Generate the artwork with circular shape (organic ink blot look)
fig = sn.plot_negative_space_artwork(
    fname="../../2025-Statistics-Masters-Thesis/figures/artwork/negative_space_artwork.png",
    figsize=(20, 20),  # Figure size in cm
    dpi=300,  # High resolution for thesis quality
    dot_size=0.5,  # Small dots for ink effect
    min_distance_kpc=0.05,  # Keep dots 0.3 kpc away from supernovae
    num_candidate_dots=2000000,  # Sample 100,000 candidate positions
    feather_radius_kpc=30,  # Feathering starts at 26 kpc
    feather_strength=5,  # Smooth exponential fade (1.0-2.0 is good range)
    shape="circle",  # Circular boundary like an ink blot
    show=True,  # Display the result
    transparent=True,  # Transparent background for thesis
    seed=42  # Reproducible randomness
)

print("✓ Artwork generated successfully")
```
