```python
from starccato_flow.supernovae.supernovae import Supernovae
from starccato_flow.plotting.sky import plot_galactic_supernovae_polar_hemispheres
```

### Configs

```python
font_family = "serif"
font_name = "Times New Roman"
```

```python
supernovae_simple = Supernovae(complex=False)
supernovae_complex = Supernovae(complex=True)
```

### R vs. Galactic Supernovae Surface Density Plot

```python
supernovae_simple._plot_surface_density(
    fname="../../2025-Statistics-Masters-Thesis/figures/supernovae/supernovae_surface_density_simple.pdf",
    font_family=font_family,
    font_name=font_name,
    transparent=False
)
```

### Simple Galactic Supernovae Distribution

```python
supernovae_simple.plot_galactic_distribution(
    fname_xy="../../2025-Statistics-Masters-Thesis/figures/supernovae/supernovae_galactic_distribution_simple.pdf",
    font_family=font_family,
    font_name=font_name,
    transparent=False,
    background="white"
)
```

### Complex Galactic Supernovae Distribution (what we end up using)

```python
supernovae_complex.plot_galactic_distribution(
    fname_xy="../../2025-Statistics-Masters-Thesis/figures/supernovae/supernovae_galactic_distribution_complex.pdf",
    font_family=font_family,
    font_name=font_name,
    transparent=False,
    background="white"
)
```

### Curriculum Learning -- Sampled Supernovae

```python
supernovae_complex.sample_supernovae_for_epoch(
    epoch=0,
    n_samples=18000,
    num_epochs=256,
    exponential=True,
    fname="../../2025-Statistics-Masters-Thesis/figures/supernovae/first_sampled_supernovae_galactic_distribution_complex.pdf",
    font_family=font_family,
    font_name=font_name,
    background="white"
)

supernovae_complex.sample_supernovae_for_epoch(
    epoch=256,
    n_samples=18000,
    num_epochs=256,
    exponential=True,
    fname="../../2025-Statistics-Masters-Thesis/figures/supernovae/final_sampled_supernovae_galactic_distribution_complex.pdf",
    font_family=font_family,
    font_name=font_name,
    background="white"
)
```

```python
ccsn = Supernovae(complex=True)  # Use a rotation offset of 90 degrees for the galactic plane

plot_galactic_supernovae_polar_hemispheres(
    ccsn,
    show_constellation_borders=True,
    example=True,
    format="thesis",
    fname="../../2025-Statistics-Masters-Thesis/figures/supernovae/supernovae_galactic_distribution_sky_plot.pdf",
    transparent=False,
    background="white",
    font_family="serif",
    coastline=True,
    n_background_supernovae=20000,
    galaxy=True,
    constellations=False,
    font_name="Times New Roman",
)
```
