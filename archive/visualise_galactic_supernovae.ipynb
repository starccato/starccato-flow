```python
from pathlib import Path
from PIL import Image
import bilby.gw.detector as det
import numpy as np
from starccato_flow.data.s_theta_old import CCSNData
from starccato_flow.utils.defaults import TEN_KPC
from starccato_flow.plotting.signals import plot_detector_signal_channels

# --------------------------
# Signal setup (same as your detector-projection cell)
# --------------------------
ccsn_data = CCSNData()
h_plus = ccsn_data.signals.T[901] / TEN_KPC
h_cross = np.zeros_like(h_plus)

ra = np.deg2rad(100)
dec = np.deg2rad(-30)
psi = 0.0
gps = 1457654242

ifos = [
    det.get_empty_interferometer('H1'),
    det.get_empty_interferometer('L1'),
    det.get_empty_interferometer('V1')
]

fs = 4096.0
t = np.arange(len(h_plus)) / fs

# Fixed delays and antenna factors for this sky position
dts = np.array([ifo.time_delay_from_geocenter(ra, dec, gps) for ifo in ifos])
dt_min = dts.min()
relative_dts = dts - dt_min
fp_fc = [
    (
        ifo.antenna_response(ra, dec, gps, psi, mode='plus'),
        ifo.antenna_response(ra, dec, gps, psi, mode='cross')
    )
    for ifo in ifos
]

# Distance bins aligned with galactic GIF timing/range
distance_bins = list(range(1, 21))  # 1..20 kpc shells
duration_ms = 280

# Use shell midpoint distance for scaling: [r, r+1) -> r+0.5
distance_midpoints = np.array([r + 0.5 for r in distance_bins])

# Reference distance for waveform is 10 kpc
d_ref = 10.0

# --------------------------
# Build GIF using plot_detector_signal_channels
# --------------------------
plots_dir = Path('plots')
frames_dir = plots_dir / 'gif_frames_signal_kpc_chunks'
frames_dir.mkdir(parents=True, exist_ok=True)
gif_path = plots_dir / 'ccsn_detector_signal_distance_1_to_20kpc.gif'

frame_paths = []
for r, d_mid in zip(distance_bins, distance_midpoints):
    scale = d_ref / d_mid

    strains = []
    for dt_rel, (fp, fc) in zip(relative_dts, fp_fc):
        h_plus_shifted = np.interp(t - dt_rel, t, h_plus, left=0, right=0)
        h_cross_shifted = np.interp(t - dt_rel, t, h_cross, left=0, right=0)
        h_signal = scale * (fp * h_plus_shifted + fc * h_cross_shifted)
        strains.append(h_signal)

    signals_3ch = np.stack(strains, axis=0)
    frame_file = frames_dir / f'frame_{r:02d}.png'

    # Use the exact plotting style helper for detector-channel frames
    plot_detector_signal_channels(
        signals=signals_3ch,
        max_value=1.0,
        detector_labels=[ifo.name for ifo in ifos],
        fname=str(frame_file),
        generated=False,
        background='black',
    )
    frame_paths.append(frame_file)

images = [Image.open(fp) for fp in frame_paths]
images[0].save(
    gif_path,
    save_all=True,
    append_images=images[1:],
    duration=duration_ms,
    loop=0,
    optimize=False,
    disposal=2,
 )
for im in images:
    im.close()

print(f'✓ Saved GIF: {gif_path}')
print(f'✓ Frames: {len(frame_paths)} (distance bins 1-2 ... 20-41 kpc)')
print(f'✓ Frame duration: {duration_ms} ms (matches galactic-map timing)')
```
